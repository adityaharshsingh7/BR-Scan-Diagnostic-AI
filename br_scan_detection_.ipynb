```python
import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image
import torch.nn as nn

# 1.ResNet-50 Model
resnet50 = models.resnet50(pretrained=True)
resnet50.eval() # Set to evaluation mode

# 2. Image Preprocessing
preprocess = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

def get_medical_score(image_path):
    try:
        input_image = Image.open(image_path).convert('RGB')
        input_tensor = preprocess(input_image)
        input_batch = input_tensor.unsqueeze(0)

        with torch.no_grad():
            output = resnet50(input_batch)

        # We take the raw output and turn it into a probability
        probabilities = torch.nn.functional.softmax(output[0], dim=0)

        score = float(torch.max(probabilities)) * 100
        return score
    except Exception as e:
        return f"Error: {e}"

healthy_path = '/content/image1'
cancer_path = '/content/image2'

healthy_score = get_medical_score(healthy_path)
cancer_score = get_medical_score(cancer_path)

print("-" * 30)
print("D3S HEALTHCARE: RESNET-50 DEEP ANALYSIS")
print("-" * 30)
print(f"HEALTHY SAMPLE SCORE    : {healthy_score:.2f}% Normal Tissue Confidence")
print(f"CANCEROUS SAMPLE SCORE  : {cancer_score:.2f}% Malignancy Probability")
print("-" * 30)
```
